```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("SalesPipeline")
    .master("local[*]")
    .getOrCreate()
)

print(spark.version)
```

```python
sales_data = [
    (1, 101, "Laptop", "Electronics", 800, 1),
    (2, 102, "Phone", "Electronics", 500, 2),
    (3, 101, "Mouse", "Accessories", 25, 3),
    (4, 103, "Keyboard", "Accessories", 50, 2),
    (5, 104, "Monitor", "Electronics", 300, 1),
    (6, 102, "USB Cable", "Accessories", 10, 5),
    (7, 105, "Laptop", "Electronics", 800, 1),
    (8, 103, "Mouse", "Accessories", 25, 2)
]

columns = [
    "order_id",
    "customer_id",
    "product",
    "category",
    "price",
    "quantity"
]

sales = spark.createDataFrame(sales_data, columns)

sales.show()
```

```python
sales.select(
    "product",
    "price",
    "quantity"
).show()
```

```python
sales.filter(sales.price > 100).show()
```

```python
from pyspark.sql.functions import col

sales = sales.withColumn(
    "total_sales",
    col("price") * col("quantity")
)

sales.show()
```

```python
from pyspark.sql.functions import sum

sales.groupBy("product").agg(
    sum("total_sales").alias("revenue")
).show()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

# Start Spark
spark = (
    SparkSession.builder
    .appName("Colab Spark ETL")
    .master("local[*]")
    .getOrCreate()
)

# Create the data
data = [
    ("pink", 10),
    ("blue", 9),
    ("green", 8),
    ("gold", 7),
    ("black", 6),
    ("pink", 7),
    ("blue", 6)
]

# Create the missing DataFrame
df = spark.createDataFrame(data, ["colour", "value"])

print("Original data:")
df.show()

# Transform the data
result = (
    df.groupBy("colour")
      .agg(F.sum("value").alias("total_value"))
      .orderBy("colour")
)

print("Transformed data:")
result.show()
```

```python
dag_code = '''
from datetime import datetime

from airflow.sdk import DAG
from airflow.providers.standard.operators.python import PythonOperator


def extract():
    print("Extracting colour data")
    return [
        ("pink", 10),
        ("blue", 9),
        ("pink", 7),
        ("blue", 6)
    ]


def transform():
    data = [
        ("pink", 10),
        ("blue", 9),
        ("pink", 7),
        ("blue", 6)
    ]

    totals = {}

    for colour, value in data:
        totals[colour] = totals.get(colour, 0) + value

    print("Transformed data:", totals)


def load():
    print("Data loaded successfully")


with DAG(
    dag_id="colour_etl_pipeline",
    start_date=datetime(2026, 1, 1),
    schedule=None,
    catchup=False,
    tags=["colab", "etl"],
) as dag:

    extract_task = PythonOperator(
        task_id="extract",
        python_callable=extract
    )

    transform_task = PythonOperator(
        task_id="transform",
        python_callable=transform
    )

    load_task = PythonOperator(
        task_id="load",
        python_callable=load
    )

    extract_task >> transform_task >> load_task
'''

with open("/content/airflow/dags/colour_etl_pipeline.py", "w") as file:
    file.write(dag_code)

print("DAG created successfully")
```
